```python
# CELL 1: Kaggle Environment & Auto-YAML
import os
import yaml
from glob import glob

print("Installing Ultralytics...")
!pip install ultralytics -q
import torch

print(f"GPU Assigned: {torch.cuda.get_device_name(0)}")

# Kaggle unzips everything automatically. This finds exactly where it put the images.
try:
    base_path = glob('/kaggle/input/**/images/train', recursive=True)[0].split('/images/train')[0]
    print(f"Dataset successfully located at: {base_path}")
    
    # Generate the Kaggle-specific YAML
    kaggle_yaml_content = {
        "path": base_path, 
        "train": "images/train",
        "val": "images/val",
        "test": "images/test",
        "names": {
            0: "Pothole",
            1: "Speedbump",
            2: "Crack"
        }
    }

    # Save it to the working directory where we have write-access
    with open('/kaggle/working/road_anomaly.yaml', 'w') as f:
        yaml.dump(kaggle_yaml_content, f, sort_keys=False)
    print("✅ Perfect Kaggle YAML map successfully created!")

except IndexError:
    print("❌ Could not find the images folder. Check if the dataset is attached correctly!")
```

```python
# CELL 2: Training Run (LR = 0.001)
from ultralytics import YOLO

print(f"{'='*50}")
print("🚀 IGNITING KAGGLE TRAINING RUN 3: Learning Rate = 0.001")
print(f"{'='*50}\n")

model = YOLO("yolov3u.pt") 

results = model.train(
    data="/kaggle/working/road_anomaly.yaml",
    epochs=100,                
    imgsz=640,                 
    batch=16,                  
    workers=4,                 
    
    optimizer="SGD",           
    lr0=0.001,                 # THE NEW LEARNING RATE 
    
    mosaic=1.0,                
    mixup=0.5,                 
    hsv_h=0.0, hsv_s=0.0, hsv_v=0.0, degrees=0.0, translate=0.0, 
    scale=0.0, fliplr=0.0, flipud=0.0,
    
    project="/kaggle/working/Road_Anomaly_YOLOv3", 
    name="Run_LR_0.001"        # RENAMED FOR CLARITY
)

print("✅ Run 3 (LR=0.001) Successfully Completed!")
```
